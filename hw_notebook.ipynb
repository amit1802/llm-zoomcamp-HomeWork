```python
from gitsource import GithubRepositoryDataReader

reader = GithubRepositoryDataReader(
    repo_owner="DataTalksClub",
    repo_name="llm-zoomcamp",
    commit_id="8c1834d",
    allowed_extensions={"md"},
    filename_filter=lambda path: "/lessons/" in path,
)

files = reader.read()

documents = []

for file in files:
    doc = file.parse()
    documents.append(doc)

```

```python
#print(documents[0].keys())
#print(type(documents[0]))
#print(documents[:2])

print("Q1. How many lesson pages are in the dataset?")
print("A1. ",len(documents))
```

```python
from minsearch import Index

index = Index(
    text_fields=["content"],
    keyword_fields=["filename"],
)
index.fit(documents)
result = index.search("How does the agentic loop keep calling the model until it stops?")
print("Q2. What's the filename of the first result?")
print("A2. ",result[0]["filename"])
```
